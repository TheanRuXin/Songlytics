```python
interest_feature_cols = ['duration_ms', 'explicit', 'danceability', 'energy', 'speechiness',
       'acousticness', 'instrumentalness', 'liveness', 'valence', 'loudness',
       'tempo']
```

```python
for feature_col in interest_feature_cols:
    pos_data = df[df["popularity_class"] == 1][feature_col]
    neg_data = df[df["popularity_class"] == 0][feature_col]
    very_popular_data = df[df["popularity_class"] == 2][feature_col]

    plt.figure(figsize=(12,7))
    sns.histplot(pos_data, bins=30, label="Popular", color="green", alpha=0.3, stat="density")
    sns.histplot(neg_data, bins=30, label="Not Popular", color="red", alpha=0.3, stat="density")
    sns.histplot(very_popular_data, bins=30, label="Very Popular", color="blue", alpha=0.3, stat="density")
    plt.legend(loc='upper right')
    plt.title(f"Histogram Plot For {feature_col}")
    plt.show()
```

```python
for feature_col in interest_feature_cols:
    plt.figure(figsize=(12, 7))
    
    sns.histplot(data=df, x=feature_col, hue="popularity_class", 
                 bins=30, alpha=0.3, stat="density",
                 palette={0: "red", 1: "green", 2: "blue"})
    
    sns.kdeplot(data=df, x=feature_col, hue="popularity_class",
                linewidth=2.5, palette={0: "darkred", 1: "darkgreen", 2: "darkblue"})
    
    plt.title(f"Distribution of {feature_col} by Popularity")
    plt.show()
```

```python
plt.figure(figsize=(12, 10))

corr_features = ['danceability', 'energy', 'speechiness', 'acousticness', 
                 'instrumentalness', 'liveness', 'valence', 'loudness', 'tempo']

corr_matrix = df[corr_features].corr()

sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, square=True, linewidths=0.5)

plt.title("Feature Correlation Heatmap", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
for feature_col in interest_feature_cols:
    plt.figure(figsize=(12, 7))
    sns.boxplot(data=df, x="popularity_class", y=feature_col,
                hue="popularity_class", legend=False,  
                palette={0: "red", 1: "green", 2: "blue"})  
    plt.title(f"Box Plot of {feature_col} by Popularity Class", fontsize=14)
    plt.xlabel("Popularity Class (0=Not Popular, 1=Popular, 2=Very Popular)", fontsize=12)
    plt.ylabel(feature_col, fontsize=12)
    plt.show()
```
