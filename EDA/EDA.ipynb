```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="popularity",
    bins=20
)

plt.xlabel("Popularity")
plt.ylabel("Number of Songs")
plt.title("Distribution of Song Popularity")
plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(
    x=df["popularity"]
)

plt.xlabel("Popularity")
plt.title("Boxplot of Song Popularity")
plt.show()
```

```python
audio_features = [
    "danceability",
    "energy",
    "loudness",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo"
]

df[audio_features].hist(
    figsize=(15, 12),
    bins=20
)

plt.suptitle("Distribution of Audio Features")
plt.tight_layout()
plt.show()
```

```python
top_genres = df.groupby("track_genre")["popularity"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index,
    palette="viridis" 
)

plt.xlabel("Average Popularity")
plt.ylabel("Genre")
plt.title("Top 10 Most Popular Genres on Average")
plt.tight_layout() 
plt.show()
```

```python
plt.figure(figsize=(12, 10))

corr_features = ['popularity','duration_ms','danceability', 'energy', 'speechiness', 'acousticness', 
                 'instrumentalness', 'liveness', 'valence', 'loudness', 'tempo']

corr_matrix = df[corr_features].corr()

sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, square=True, linewidths=0.5)

plt.title("Feature Correlation Heatmap", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
popularity_corr = corr_matrix["popularity"].sort_values(ascending=False)

print(popularity_corr)
```
